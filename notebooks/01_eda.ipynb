# Missing Values Analysis

Missing values are important to identify before training a machine learning model because many algorithms cannot handle them directly. Understanding their amount and location helps guide later decisions about data preparation, while leaving the original data unchanged at this stage.

```python
missing_counts = df.isna().sum()

if missing_counts.sum() == 0:
    print("Great news: no missing values were found in the dataset.")
else:
    missing_summary = pd.DataFrame({
        "missing_count": missing_counts,
        "missing_percentage": (missing_counts / len(df) * 100).round(2),
    })
    missing_summary = missing_summary[missing_summary["missing_count"] > 0]
    missing_summary.sort_values("missing_count", ascending=False)
```

# Duplicate Records

Duplicate records can give repeated observations more influence during model training, which may bias learned patterns. They can also make validation results appear stronger than they would be on genuinely new data, so identifying them is an important step before later data preparation.

```python
duplicate_rows = df.duplicated().sum()
print(f"Duplicate rows: {duplicate_rows}")
```

# Target Variable Distribution

Checking the target-variable distribution before model training helps identify class imbalance. When one class is much less frequent than another, an accuracy-only evaluation can be misleading and the model may need special handling later.

```python
target_counts = df["Churn"].value_counts()
target_percentages = df["Churn"].value_counts(normalize=True).mul(100)

target_distribution = pd.DataFrame({
    "count": target_counts,
    "percentage": target_percentages.round(2),
})
target_distribution
```

```python
plt.figure(figsize=(6, 4))
plt.bar(target_counts.index.astype(str), target_counts.values, color="steelblue")
plt.title("Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Numerical Features Distribution

Inspecting numerical feature distributions helps reveal each variable's range, concentration, skewness, and possible outliers before model training. This understanding can inform later modeling decisions without changing the data at this stage.

```python
target_column = "Churn"
numerical_features = df.select_dtypes(include=np.number).columns.drop(target_column, errors="ignore").tolist()
numerical_features
```

```python
if not numerical_features:
    print("No numerical features were found to plot.")
else:
    n_columns = 2
    n_rows = int(np.ceil(len(numerical_features) / n_columns))
    fig, axes = plt.subplots(n_rows, n_columns, figsize=(12, 4 * n_rows))
    axes = np.atleast_1d(axes).ravel()

    for axis, feature in zip(axes, numerical_features):
        axis.hist(df[feature].dropna(), bins=30, color="steelblue", edgecolor="white")
        axis.set_title(f"Distribution of {feature}")
        axis.set_xlabel(feature)
        axis.set_ylabel("Frequency")

    for axis in axes[len(numerical_features):]:
        axis.set_visible(False)

    plt.tight_layout()
    plt.show()
```

# Outlier Detection

An outlier is an observation with a value that is unusually far from most other values in a feature. Detecting outliers matters in machine learning because extreme values can disproportionately influence model training, distort summary statistics, and affect model performance. This section only identifies potential outliers; it does not remove or modify any records.

```python
if not numerical_features:
    print("No numerical features were found to plot.")
else:
    n_columns = 2
    n_rows = int(np.ceil(len(numerical_features) / n_columns))
    fig, axes = plt.subplots(n_rows, n_columns, figsize=(12, 4 * n_rows))
    axes = np.atleast_1d(axes).ravel()

    for axis, feature in zip(axes, numerical_features):
        axis.boxplot(df[feature].dropna())
        axis.set_title(f"Boxplot of {feature}")
        axis.set_xlabel(feature)
        axis.set_ylabel("Value")

    for axis in axes[len(numerical_features):]:
        axis.set_visible(False)

    plt.tight_layout()
    plt.show()
```

# Correlation Analysis

Correlation measures the strength and direction of a linear relationship between two numerical variables. Values close to 1 or -1 indicate a strong positive or negative relationship, while values near 0 indicate little linear relationship. Highly correlated features should be investigated before model training because they may contain overlapping information, which can make some models less stable or harder to interpret. This section only examines relationships; it does not remove any features.

```python
numerical_columns = df.select_dtypes(include=np.number).columns
correlation_matrix = df[numerical_columns].corr()
correlation_matrix
```

```python
if correlation_matrix.empty:
    print("No numerical columns were found for correlation analysis.")
else:
    figure_size = max(8, len(numerical_columns) * 0.7)
    plt.figure(figsize=(figure_size, figure_size))
    image = plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)
    plt.colorbar(image, label="Correlation")
    plt.xticks(range(len(numerical_columns)), numerical_columns, rotation=90)
    plt.yticks(range(len(numerical_columns)), numerical_columns)
    plt.title("Correlation Matrix of Numerical Features")
    plt.tight_layout()
    plt.show()
```

# Categorical Feature Analysis

Categorical distributions show how frequently each group or label occurs in a feature. Inspecting them before model training can reveal rare categories, uneven group sizes, and potential data-quality issues that may affect how a model learns from those variables. This section examines the original categories only; it does not encode or modify the data.

```python
target_column = "Churn"
categorical_features = (
    df.select_dtypes(include=["object", "category", "bool"]).columns
    .drop(target_column, errors="ignore")
    .tolist()
)

if not categorical_features:
    print("No categorical features were found to analyze.")
else:
    for feature in categorical_features:
        value_counts = df[feature].value_counts()
        print(f"Value counts for {feature}:")
        display(value_counts)

        plt.figure(figsize=(8, 4))
        plt.bar(value_counts.index.astype(str), value_counts.values, color="steelblue")
        plt.title(f"Distribution of {feature}")
        plt.xlabel(feature)
        plt.ylabel("Number of Records")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()
```

# Feature Relationship with Target

Comparing each feature across target classes helps identify potentially useful predictive signals. Numerical features whose distributions differ by class and categorical features whose frequencies vary by class may help a future model distinguish outcomes. This is exploratory analysis only; no model is built in this section.

```python
target_column = "Churn"
numerical_features = df.select_dtypes(include=np.number).columns.drop(target_column, errors="ignore").tolist()
categorical_features = (
    df.select_dtypes(include=["object", "category", "bool"]).columns
    .drop(target_column, errors="ignore")
    .tolist()
)
target_classes = df[target_column].dropna().unique()
```

```python
if not numerical_features:
    print("No numerical features were found to compare with the target.")
else:
    for feature in numerical_features:
        plt.figure(figsize=(8, 4))

        for target_class in target_classes:
            values = df.loc[df[target_column] == target_class, feature].dropna()
            plt.hist(values, bins=30, alpha=0.5, label=str(target_class), edgecolor="white")

        plt.title(f"{feature} Distribution by {target_column}")
        plt.xlabel(feature)
        plt.ylabel("Frequency")
        plt.legend(title=target_column)
        plt.tight_layout()
        plt.show()
```

```python
if not categorical_features:
    print("No categorical features were found to compare with the target.")
else:
    for feature in categorical_features:
        category_target_counts = pd.crosstab(df[feature], df[target_column])
        print(f"Category frequencies for {feature} by {target_column}:")
        display(category_target_counts)

        positions = np.arange(len(category_target_counts.index))
        bar_width = 0.8 / len(category_target_counts.columns)
        plt.figure(figsize=(max(8, len(category_target_counts.index) * 0.7), 4))

        for index, target_class in enumerate(category_target_counts.columns):
            offset = (index - (len(category_target_counts.columns) - 1) / 2) * bar_width
            plt.bar(
                positions + offset,
                category_target_counts[target_class],
                width=bar_width,
                label=str(target_class),
            )

        plt.title(f"{feature} Frequencies by {target_column}")
        plt.xlabel(feature)
        plt.ylabel("Number of Records")
        plt.xticks(positions, category_target_counts.index.astype(str), rotation=45, ha="right")
        plt.legend(title=target_column)
        plt.tight_layout()
        plt.show()
```
